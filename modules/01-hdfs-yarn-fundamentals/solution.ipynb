# 01 - HDFS and YARN fundamentals - Solution

Опорный сценарий: загрузка файла в HDFS и чтение его Spark-приложением через YARN.


```python
reference_commands = [
    "hdfs dfs -mkdir -p /data/course/orders",
    "hdfs dfs -put -f /workspace/dataset/orders.csv /data/course/orders/",
    "hdfs dfs -ls /data/course/orders",
    "hdfs dfs -du -h /data/course/orders",
]
reference_commands
```

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("lesson01_hdfs_yarn")
    .master("yarn")
    .config("spark.submit.deployMode", "client")
    .getOrCreate()
)

orders_df = spark.read.option("header", True).csv(
    "hdfs://namenode:8020/data/course/orders/orders.csv"
)
print("rows:", orders_df.count())
orders_df.printSchema()
orders_df.show(5, truncate=False)
```
