# 03 - Airflow basic - Solution


```python
from datetime import datetime, timedelta
from airflow import DAG
from airflow.operators.empty import EmptyOperator
from airflow.operators.python import PythonOperator


def print_context(**context):
    print("run_id:", context["run_id"])
    print("logical_date:", context["logical_date"])


with DAG(
    dag_id="lesson03_basic_dag",
    start_date=datetime(2024, 1, 1),
    schedule="@daily",
    catchup=False,
    default_args={"owner": "course", "retries": 1, "retry_delay": timedelta(minutes=1)},
) as dag:
    start = EmptyOperator(task_id="start")
    echo_context = PythonOperator(task_id="echo_context", python_callable=print_context)
    finish = EmptyOperator(task_id="finish")
    start >> echo_context >> finish
```
